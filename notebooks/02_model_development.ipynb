# 02 - Primary and challenger models

Development uses 2022-2024. The 2025 vintage is held out for out-of-time validation.

```python
from pathlib import Path
import sys
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.data import generate_portfolio, FEATURES, TARGET
from src.model import build_logistic_model, build_challenger_model
df=generate_portfolio(); train=df[df.observation_year<=2024]; test=df[df.observation_year==2025]
primary=build_logistic_model().fit(train[FEATURES],train[TARGET])
challenger=build_challenger_model().fit(train[FEATURES],train[TARGET])
print('Development rows:',len(train),'OOT rows:',len(test))
```

```python
from src.validation import evaluate_model
for name, model in [('Logistic', primary), ('Decision tree', challenger)]:
    p=model.predict_proba(test[FEATURES])[:,1]
    print(name, evaluate_model(test[TARGET],p))
```
