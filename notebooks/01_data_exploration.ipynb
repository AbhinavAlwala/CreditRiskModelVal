# 01 - Portfolio exploration

The portfolio is synthetic and reproducible. This notebook checks its shape, target rate, exposure profile, and major risk-driver relationships before modelling.


```python
from pathlib import Path
import sys, pandas as pd
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
from src.data import generate_portfolio
df = generate_portfolio()
df.head()
```

```python
df.groupby('observation_year')['default_12m'].agg(['count','mean']).rename(columns={'mean':'default_rate'})
```

```python
df[['annual_income','debt_to_income','credit_utilization','delinquencies_2y','inquiries_6m','default_12m']].corr().round(3)
```

```python
df.groupby('observation_year').agg(borrowers=('customer_id','size'), total_ead=('exposure_at_default','sum'), avg_lgd=('loss_given_default','mean'), default_rate=('default_12m','mean')).round(3)
```
